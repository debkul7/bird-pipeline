```python
import os
from dotenv import load_dotenv

load_dotenv()
api_key = os.getenv("EBIRD_API_KEY") 
print("API key loaded:", api_key is not None)
```

```python
import requests

url = "https://api.ebird.org/v2/data/obs/geo/recent/notable"

headers = {"X-eBirdApiToken": api_key}

params = {
    "lat": 51.11,    # Wrocław center
    "lng": 17.03,
    "dist": 50,      # distance radius in km (max in eBird is 50)
    "back": 30,      # how many days back (max is 30)
}

response = requests.get(url, headers=headers, params=params)

print("Status:", response.status_code)
```

```python
data = response.json()

print("Data type:", type(data))
print("Observation list:", len(data))
```

```python
import pandas as pd

df = pd.DataFrame(data)

print(df.shape)
df.head()
```

```python
import duckdb

duckdb.sql("""
    SELECT comName, 
        COUNT(*) AS observations, 
        COUNT(DISTINCT locName) AS places_seen
    FROM df
    GROUP BY comName
    ORDER BY observations DESC
""")
```
